```python
### Plots for Nico:

def custom_histogram_lumclass_1(skiff_values, simbad_values, ax):

  labels = ['I', 'II', 'III', 'IV', 'V', 'VI']
  bin_edges = [i for i in range(0, 7)]

  skiff_counts = ax.hist(skiff_values, bins=bin_edges, edgecolor='black', label=f'Skiff (n = {skiff_values.shape[0]})', alpha=0.5)[0]
  simbad_counts = ax.hist(simbad_values, bins=bin_edges, edgecolor='black', label=f'SIMBAD (n = {simbad_values.shape[0]})', alpha=0.5)[0]

  for j in range(0, 6):
    ax.text(bin_edges[j] + (bin_edges[j + 1] - bin_edges[j]) / 2, -600, labels[j], ha='center', va='bottom')
    ax.text(bin_edges[j] + (bin_edges[j + 1] - bin_edges[j]) / 2, 300, f'Skiff = {int(skiff_counts[j])}\nSIMBAD = {int(simbad_counts[j])}', ha='center', va='bottom', rotation='vertical')

  ax.set_xlabel('Luminosity Class', labelpad=18)
  ax.set_ylabel('n')
  ax.legend()
  ax.set_xticks([])
  #ax.set_yscale('log')
  #ax.set_yticks([1, 10, 100, 1000, 10000, 100000])

simbad_condition = (simbad["lum_codification"].astype(str).str[-2:] !=".5")
simbad_condition = (simbad_condition & ~simbad['is_OB'])
simbad_values = simbad['lum_codification'][simbad_condition].values
skiff_condition = (skiff['lum_codification_mean'].notna()) & (skiff["lum_codification_mean"].astype(str).str[-2:] !=".5")
skiff_values = skiff.loc[skiff_condition]['lum_codification_mean'].values

fig, ax = pl.subplots(1)

custom_histogram_lumclass_1(skiff_values, simbad_values, ax)
ax.set_title('Distribution of Lum. Classes\nSkiff vs. SIMBAD')
fig.savefig(f'LumClass_Distribution_Without_OB_for_Skiff_StDev_3.png')
fig.show()
```
