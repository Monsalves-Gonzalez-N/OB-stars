```python
from astroquery.simbad import Simbad
from astropy.coordinates import SkyCoord
import astropy.units as u
import pandas as pd
import numpy as np
from time import sleep

# Agregar el campo 'sptype' a la consulta
Simbad.add_votable_fields('sptype')

# Definir radio de búsqueda
radius = 2.0 * u.arcsec

# Número de objetos en cada batch (SIMBAD puede aceptar un límite, prueba con 5000-10000)
batch_size = 10  

# Extraer coordenadas
ra_values = OBAGDR3["ra"].values
dec_values = OBAGDR3["dec"].values

# Dividir en lotes
ra_batches = np.array_split(ra_values, len(ra_values) // batch_size)
dec_batches = np.array_split(dec_values, len(dec_values) // batch_size)

# Lista para almacenar los resultados
all_results = []

for ra_batch, dec_batch in zip(ra_batches, dec_batches):
    try:
        # Convertir a SkyCoord
        c = SkyCoord(ra=ra_batch * u.deg, dec=dec_batch * u.deg)
        
        # Hacer la consulta para este lote
        tab = Simbad.query_region(c, radius=radius)
        
        # Almacenar resultados
        if tab is not None:
            all_results.append(tab.to_pandas())  # Convertir a Pandas y guardar
        
        # Pausa para evitar bloquear SIMBAD
        sleep(1)  # Evita sobrecargar el servidor de SIMBAD
        
    except Exception as e:
        print(f"Error en el batch: {e}")
        continue

# Unir todos los resultados en un solo DataFrame
if all_results:
    final_df = pd.concat(all_results, ignore_index=True)
    import ace_tools as tools
    tools.display_dataframe_to_user(name="SIMBAD Query Results", dataframe=final_df)
```
