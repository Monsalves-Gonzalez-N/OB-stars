```python
import numpy as np
import astroquery
from astropy import units as u
from astroquery.simbad import Simbad
from astropy.coordinates import SkyCoord
import pandas as pd
from astropy.table import Table, vstack
```

```python
radius=2.
df_2 = pd.read_csv('/home/nicolas/nico/Data/Masivas/Data_OB_stars/OBAGDR3.csv')
```

```python
df_2 = df_2.head(100)
```

```python
Simbad.TIMEOUT = 100000
Simbad.ROW_LIMIT = 0
```

```python
Simbad.add_votable_fields('sptype')
Simbad.add_votable_fields('sp_bibcode')
coordinates = SkyCoord(df_2[["ra","dec"]].values,unit=("deg","deg"))
tabF = Simbad.query_region(coordinates, radius=radius*u.arcsec)
```

```python
table=tabF.to_pandas()
```
